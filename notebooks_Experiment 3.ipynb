```python
#Experiment 3.a
import matplotlib.pyplot as cricket
Overs=list(range(5,51,5))
Australia_Score=[35,60,95,130,170,205,245,280,320,360]
England_Score=[20,50,85,115,150,180,210,250,290,330]
cricket.plot(Overs,Australia_Score)
cricket.plot(Overs,England_Score)
cricket.title("AUSTRALIA Vs ENGLAND")
cricket.xlabel("Overs")
cricket.ylabel("Score")
cricket.plot(Overs,Australia_Score,color="green",label="AUSTRALIA")
cricket.plot(Overs,England_Score,color="red",label="ENGLAND")
cricket.legend(loc="center right")
cricket.show()
```

```python

#experiment 3.b
import matplotlib.pyplot as hscmark
import numpy as np
Names = ['HARISH', 'KIRUBA', 'MATHAVAN ', 'MARI', 'MONISH', 'KARTHI']
xaxis = np.arange(len(Names))
Percentage_hsc = [89, 95, 87, 78, 68, 92]
hscmark.bar(Names, Percentage_hsc)
hscmark.xticks(xaxis, Names, rotation=45)
hscmark.xlabel("Names of Pupil")
hscmark.ylabel("Percentage")
hscmark.title("Comparison of HSC Percentage", fontsize=20, color="green")
hscmark.show()
```

```python
#EExperiment 3.c
import matplotlib.pyplot as election
# Election data
labels = ['ARUN', 'KAVIN', 'PRIYA', 'DIVYA']
Votes = [280, 190, 320, 210]
colors = ['blue', 'purple', 'gold', 'cyan']
explode = (0.2, 0, 0, 0)
# Plotting the pie chart
election.pie(Votes, labels=labels, colors=colors, explode=explode, autopct='%0.2f%%')
election.title('College Election Results')
election.show()
```
